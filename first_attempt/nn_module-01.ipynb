```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
#classification model
class ClassificationModel_one(nn.Module):
    def __init__(self, num_features):
        super(ClassificationModel_one   , self).__init__()
        self.linear1 = nn.Linear(num_features, 3)
        self.relu= nn.ReLU()
        self.linear2 = nn.Linear(3, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self,x):
        out = self.linear1(x)
        out = self.relu(out)
        y_pred = self.sigmoid(self.linear2(out))
        return y_pred
    
```

```python
features=torch.rand(10,5)

model=ClassificationModel_one(num_features=features.shape[1])
y_pred=model(features)
y_pred
```

```python
from torchinfo import summary
summary(model, input_size=(10,5))
```

```python
#using nn.Sequential to create the same model
class ClassificationModel_two(nn.Module):
    def __init__(self, num_features):
        super(ClassificationModel_two, self).__init__()
        self.model = nn.Sequential(
            nn.Linear(num_features, 3),
            nn.ReLU(),
            nn.Linear(3, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        out=self.model(x)
        return out
```

```python
model=ClassificationModel_two(num_features=features.shape[1])
y_pred=model(features)
y_pred
```

```python
summary(model, input_size=(10,5))
```
